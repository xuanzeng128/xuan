## 作業說明
整理桌遊體驗與電影欣賞兩場活動的報名資料，練習字串方法、f-string，以及 list、tuple、dict、set 的搭配使用。

- `tuple`：保存固定的兩個活動名稱。
- `list`：保存清理後仍保留重複項目的名單，以及排序後的名單。
- `dict`：查詢活動單價並保存人數、報名人次與應收摘要。
- `set`：去除同場重複報名，並進行交集、聯集與差集運算。

桌遊每人 120 元，電影每人 80 元；每場各自去重後計費。以下三組測試均重新執行完整程式。

## 測試一：原始資料

電影原始名單：` ben, Dora, ERIC, ben `



```python
# 固定活動名稱與原始資料
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 從 tuple 讀取活動名稱
board_name, movie_name = activity_names

# 清理字串並拆成 list；保留重複項目
board_clean_text = board_raw.replace(" ", "").lower()
movie_clean_text = movie_raw.replace(" ", "").lower()
board_list = board_clean_text.split(",")
movie_list = movie_clean_text.split(",")

# 轉成 set 去重
board_set = set(board_list)
movie_set = set(movie_list)

# 集合運算：交集、聯集、兩個方向的差集
common_set = board_set & movie_set
all_set = board_set | movie_set
board_only_set = board_set - movie_set
movie_only_set = movie_set - board_set

# 排序後轉成 list，並切片取得前兩位
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
common_sorted = sorted(common_set)
all_sorted = sorted(all_set)
board_only_sorted = sorted(board_only_set)
movie_only_sorted = sorted(movie_only_set)
notice_preview = all_sorted[:2]

# 查詢單價並建立摘要字典
board_fee = unit_fees[board_name]
movie_fee = unit_fees[movie_name]
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(common_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * board_fee,
    "電影應收": len(movie_set) * movie_fee,
    "總應收": len(board_set) * board_fee + len(movie_set) * movie_fee
}

# 輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{board_name}，每人 {board_fee:.2f} 元")
print(f"活動：{movie_name}，每人 {movie_fee:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notice_preview}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```

## 個人回答

### 問題一
測試一的桌遊 list 有 4 筆，是因為 Amy 在原始名單中出現兩次。list 會保留每筆資料及原本順序，因此能看出重複報名；但若直接用 list 長度計費，就會重複收取 Amy 的費用。轉成 set 後，重複姓名會合併，只留下 amy、ben、cora 三個不同的人，所以桌遊以 3 人計費。set 適合去重與集合比較，但不保證原本順序，也不保留重複次數。

### 問題二
測試三中，桌遊與電影去重後的集合完全相同，都是 {'amy', 'ben', 'cora'}。因此交集有 3 人、兩個方向的差集都是空集合，聯集也只有 3 人。總不重複人數是 3 人；每個人都參加兩場，所以報名人次為 3 + 3 = 6 人次。費用分場計算：桌遊 3 × 120 = 360 元，電影 3 × 80 = 240 元，總應收為 360 + 240 = 600 元。
